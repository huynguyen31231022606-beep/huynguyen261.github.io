#### Biểu thức tính toán

```python
# Nhập dữ liệu
a = int(input("Moi ban nhap so a: "))
b = int(input("Moi ban nhap so b: "))
# Xử lý
kqCong = a + b
kqTru = a - b
kqNhan = a * b
kqChiaNguyen = a // b
kqDu= a % b
kqChiaThuc = a / b
# Xuất dữ liệu
print("Cac ket qua tinh toan: ")
print("%-5d + %5d = %5d"%(a, b, kqCong))
print("%-5d - %5d = %5d"%(a, b, kqTru))
print("%-5d * %5d = %5d"%(a, b, kqNhan))
print("%-5d / %5d = %5d"%(a, b, kqChiaNguyen))
print("%-5d / %5d = %5d"%(a, b, kqDu))
print("%-5d / %5d = %5.2f"%(a, b, kqChiaThuc))
```

#### Biểu thức so sánh

```python
# Nhập dữ liệu
a = float(input("Moi ban nhap so a: "))
b = float(input("Moi ban nhap so b: "))
# Xử lý
laNhoHon = (a < b)
laNhoHonBang = (a <= b)
laLonHon = (a > b)
laLonHonBang = (a >= b)
laBang = (a == b)
laKhacNhau = (a != b)
# Xuất dữ liệu
print("Ket qua so sanh hai so thuc %.2f va %.2f:"%(a, b))
print("%-8.2f < %8.2f: %8s"%(a, b, laNhoHon))
print("%-8.2f <= %8.2f: %8s"%(a, b, laNhoHonBang))
print("%-8.2f > %8.2f: %8s"%(a, b, laLonHon))
print("%-8.2f >= %8.2f: %8s"%(a, b, laLonHonBang))
print("%-8.2f == %8.2f: %8s"%(a, b, laBang))
print("%-8.2f != %8.2f: %8s"%(a, b, laKhacNhau))
```

#### Tính giá trị hàm số

```python
import math

# Nhập dữ liệu
x = float(input("Moi ban nhap vao gia tri cua bien so x: "))

# Xử lý
if x == 0:
    print("Khong the tinh f(x) vi x = 0 (chia cho 0).")
else:
    # 5! = 120
    term1 = x
    term2 = math.pow(x, 5) / math.factorial(5)
    
    # xử lý an toàn cho phần sqrt(|x|)/x^(3/2)
    if x > 0:
        term3 = math.sqrt(x) / math.pow(x, 1.5)
    else:
        # với x âm: sqrt(|x|) / (|x|^(3/2)) * (-1)^(3/2)
        # đơn giản hoá: = 1/|x|
        term3 = 1.0 / abs(x)
    
    fx = term1 + term2 + term3
    print("Gia tri cua ham so f(%.2f) = %.2f." % (x, fx))
```

#### Tổng chữ số

```python
# Nạp thư viện
import math
# Nhập dữ liệu
x = float(input("Moi ban nhap vao gia tri cua bien so x: "))
# Xử lý
fx = x + math.pow(x, 5) / (1 * 2 * 3 * 4 * 5) + \
math.sqrt(abs(x)) / math.pow(x, 3.0 / 2)
# Xuất dữ liệu
print("Gia tri cua ham so f(%.2f) = %.2f."%(x, fx))
```

#### Bài 5: Tính diện tích tam giác

```python
import math

# Nhập độ dài 3 cạnh tam giác
a = float(input("Nhap canh a: "))
b = float(input("Nhap canh b: "))
c = float(input("Nhap canh c: "))

# Kiểm tra điều kiện tạo thành tam giác
if a + b > c and a + c > b and b + c > a:
    p = (a + b + c) / 2
    S = math.sqrt(p * (p - a) * (p - b) * (p - c))
    print("Dien tich tam giac = %.2f" % S)
else:
    print("Ba canh khong tao thanh tam giac hop le.")
```

#### Bài 9: Hàm số 1

```python
import math

x = float(input("Nhap so thuc x: "))

# Tinh y1 (chi hop le khi x >= 0)
if x < 0:
    print("y1 khong xac dinh vi can bac hai cua x voi x < 0.")
    y1 = None
else:
    y1 = 4 * (x**2 + 10*x*math.sqrt(x) + 3*x + 1)
    print("y1 = %.2f" % y1)

# Tinh y2 (mau so khac 0)
den = math.exp(2*x) + math.cos((math.pi/4) * x)
if abs(den) < 1e-12:
    print("y2 khong xac dinh vi mau so = 0.")
else:
    y2 = (math.sin(math.pi * x**2) + math.sqrt(x**2 + 1)) / den
    print("y2 = %.2f" % y2)
```

#### Bài 10: KHOẢNG CÁCH

```python
import math

def read_float(prompt):
    while True:
        s = input(prompt).strip()
        s = s.replace(',', '.')            # cho phép dùng dấu phẩy cho phần thập phân
        try:
            return float(s)
        except ValueError:
            print("Giá trị không hợp lệ. Hãy nhập một số thực (ví dụ 3.14).")

x1 = read_float("Nhập x1: ")
y1 = read_float("Nhập y1: ")
x2 = read_float("Nhập x2: ")
y2 = read_float("Nhập y2: ")

d = math.hypot(x2 - x1, y2 - y1)
print("Khoảng cách AB = %.2f" % d)
```
