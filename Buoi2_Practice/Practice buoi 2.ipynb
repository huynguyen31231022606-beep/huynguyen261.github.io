```python
#So sánh
# Nhập dữ liệu 
a = float(input("Moi ban nhap so a: ")) 
b = float(input("Moi ban nhap so b: "))
# Xử lý 
laNhoHon = (a < b) 
laNhoHonBang = (a <= b)
laLonHon = (a > b) 
laLonHonBang = (a >= b)
laBang = (a == b)
laKhacNhau = (a != b)
# Xuất dữ liệu 
print("Ket qua so sanh hai so thuc %.2f va %.2f:"%(a, b))
print("%-8.2f < %8.2f: %8s"%(a, b, laNhoHon))
print("%-8.2f <= %8.2f: %8s"%(a, b, laNhoHonBang))
print("%-8.2f > %8.2f: %8s"%(a, b, laLonHon))
print("%-8.2f >= %8.2f: %8s"%(a, b, laLonHonBang))
print("%-8.2f == %8.2f: %8s"%(a, b, laBang))
print("%-8.2f != %8.2f: %8s"%(a, b, laKhacNhau))
```

```python
#Max3Num
# Input a, b, c
a = int(input())
b = int(input())
c = int(input())

# max a, b, c
def max_fn(a, b):
    return a if a > b else b
    pass # max_fn

vmax = max_fn(a, b)
vmax = max_fn(vmax, c)

# output
print("Max(%d, %d, %d) = %d"%(a, b, c, vmax))
```

```python
x = ['apple', 'banana', 'cherry']
x.append('hello')
print(x, type(x), x[1:3])

x = ('apple', 'banana', 'cherry') # immutable
# x.append('hello')
x = ('apple', 'banana', 'cherry', 'hello')
print(x, type(x), x[1:3])

x = {'name': 'Tome', 'age': 20}   # dictionary
print(x, type(x), x['name'], x['age'])

x = {'apple', 'banana', 'cherry'} # set
x.add('hello')
print(x, type(x))

x = frozenset({'apple', 'banana', 'cherry'}) # frozenset
print(x, type(x))

'''
Chuỗi bình thường có thể thay đổi và edit thêm thông tin
append ở đây có nghĩa là thêm trường dữ liệu

Rieng dictionary và frozen set là không thể edit thêm thông tin
Chuỗi set có sắp xếp thứ tự các kí tự theo trình tự a-z
'''
```
