```python
#So sánh
# Nhập dữ liệu 
a = float(input("Moi ban nhap so a: ")) 
b = float(input("Moi ban nhap so b: "))
# Xử lý 
laNhoHon = (a < b) 
laNhoHonBang = (a <= b)
laLonHon = (a > b) 
laLonHonBang = (a >= b)
laBang = (a == b)
laKhacNhau = (a != b)
# Xuất dữ liệu 
print("Ket qua so sanh hai so thuc %.2f va %.2f:"%(a, b))
print("%-8.2f < %8.2f: %8s"%(a, b, laNhoHon))
print("%-8.2f <= %8.2f: %8s"%(a, b, laNhoHonBang))
print("%-8.2f > %8.2f: %8s"%(a, b, laLonHon))
print("%-8.2f >= %8.2f: %8s"%(a, b, laLonHonBang))
print("%-8.2f == %8.2f: %8s"%(a, b, laBang))
print("%-8.2f != %8.2f: %8s"%(a, b, laKhacNhau))
```

```python
#Max3Num
# Input a, b, c
a = int(input())
b = int(input())
c = int(input())

# max a, b, c
def max_fn(a, b):
    return a if a > b else b
    pass # max_fn

vmax = max_fn(a, b)
vmax = max_fn(vmax, c)

# output
print("Max(%d, %d, %d) = %d"%(a, b, c, vmax))
```

```python

# dictionary
accounts = {
    "1001": {"name": "Cash", "balance": 500.0, "type": "Asset"},
    "1002": {"name": "Account Receivable", "balance": 1000.0, "type": "Asset"},
    "1003": {"name": "Account Payable", "balance": 750.0, "type": "Liability"},
}

for acc_num, acc_info in accounts.items():
  print('-'*50)
  print('Account Number: ', acc_num)

  print('Account Name: ', acc_info['name'])
  print('Account Balance: ', acc_info['balance'])
  print('Account Type: ', acc_info['type'])
```

## Control logic and loops

```python
# If else
score = 80
if score >= 60:
  print('Pass')
else: # score < 60
  print('Fail')
```

```python
# if
score = 80
if score >= 90:
  print('A')
elif score >= 60:  # score < 90 and score >=60
  print('Pass')
else: # score < 60
  print('Fail')
     
```

```python
# if
score = 90
grade = ''
if score >= 90:  # [>90]
  grade = 'A'
elif score >= 80:  # score < 90 and score >=80 [80, 90)
  grade = 'B'
elif score >= 70:  # [70, 80)
  grade = 'C'
elif score >= 60:  # [60, 70)
  grade = 'D'
else: # score < 60
  grade = 'E'
print('Grade: ', grade)
```

```python
# class
class Person:
  def __init__(self, name, age): # constructor (ham tao doi tuong)
    self.name = name
    self.age = age

  def myfunc(self):
    print('Hello my name is ', self.name)
    pass

p1 = Person('Alan', 20)
print(p1.name, p1.age)
p1.myfunc()
```

```python
#class
class Vehicle:
  name = ""
  kind = "car"
  color = ""
  value = 100.0

  def __init__(self, name="", kind="car", color="", value=100.0):
    self.name = name
    self.kind = kind
    self.color = color
    self.value = value
    pass

  def description(self):
    desc_str = f'{self.name} is a {self.color} {self.kind} worth ${self.value:.2f}.'
    return desc_str
    pass

car1 = Vehicle()
car1.name = 'Fer'
car1.color = 'red'
car1.kind = 'convertible'
car1.value = 60000.0

car2 = Vehicle("Jump", 'blue', 'van', 10000.0)

print(car1.description())
print(car2.description())
```
